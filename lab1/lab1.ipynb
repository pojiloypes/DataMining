# Задание 1

Посмотреть документацию к API GitHub, разобраться как вывести список репозиториев для конкретного пользователя, сохранить JSON-вывод в файле *.json.

Используем публичные репозитории пользователя `octocat`. Для другого пользователя достаточно изменить `username`.

[Документация GitHub](https://docs.github.com/en/rest/repos/repos#list-repositories-for-a-user): `GET /users/{username}/repos`. Следующая страница указана в заголовке `Link`.

Основа: `lectures/Лекция3-4.pdf`, с. 12–13 — пагинация GitHub и сессия с повторами временных ошибок. JSON сохраняется в текущую рабочую папку.

```python
import json
import time
import requests

username = "pojiloypes"
url = f"https://api.github.com/users/{username}/repos"
repos = []

response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    for row in data:
        print(f'id = {row['id']}\t name={row['name']}')
        
    filename = f"results/{username}_repos.json"
    with open(filename, "w", encoding="utf-8") as file:
        json.dump(data, file, ensure_ascii=False, indent=2)
```

# Задание 2

Изучить список открытых API (https://www.programmableweb.com/category/all/apis).
Найти среди них любое, требующее авторизацию (любого типа). Выполнить запросы к нему, пройдя авторизацию.
Ответ сервера записать в файл.
Если нет желания заморачиваться с поиском, возьмите API вконтакте (https://vk.com/dev/first_guide).
Сделайте запрос, чтб получить список всех сообществ на которые вы подписаны.

```python
# https://www.programmableweb.com/category/all/apis Больше не работает поэтому использую ВК АПИ

print('Для получения доступа перейдите по данной ссылке и введите ссылку на странице после авторизации: "https://oauth.vk.com/authorize?client_id=51772080&display=page&redirect_uri=https://oauth.vk.com/blank.html&scope=groups&response_type=token&v=5.131"')
raw_token = input()
token = raw_token.split('access_token=')[1].split('&expires_in')[0]
#print(token)

url_base = 'https://api.vk.com/method'
url_for_id = f'{url_base}/users.get'
url_for_subs = f'{url_base}/users.getSubscriptions'

user_response = requests.get(
    f'{url_for_id}?v=5.199',
    headers={"Authorization": f"Bearer {token}"}
)

if user_response.status_code == 200:
    user_data = user_response.json()
    user_id = user_data['response'][0]['id']
    
    subs_response = requests.get(
        f'{url_for_subs}?fields=membersCount&user_id={user_id}&v=5.199&extended=1',
        headers={"Authorization": f"Bearer {token}"}
    )
    
    subs_data = subs_response.json()['response']['items']
    subs_groups = [item for item in subs_data if item.get('type') == 'page']
    
    for item in subs_groups:
        print(f'id={item['id']}\tname={item['name']}')
    
    filename = f"results/{user_id}_groups.json"
    with open(filename, "w", encoding="utf-8") as file:
        json.dump(subs_groups, file, ensure_ascii=False, indent=2)





```
